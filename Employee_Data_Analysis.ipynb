```python
import pandas as pd
import os

# ==========================================
# EMPLOYEE DATA ANALYSIS PROJECT
# ==========================================

# Name of input CSV file
INPUT_FILE = "Employee.csv"

# Name of output CSV file
OUTPUT_FILE = "employee_analysis_results.csv"


# ------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------

print("Loading employee dataset...")

try:
    df = pd.read_csv(INPUT_FILE)
    print("Dataset loaded successfully!")

except FileNotFoundError:
    print(f"Error: '{INPUT_FILE}' not found.")
    print("Please place the CSV file in the same folder as this Python file.")
    exit()


# ------------------------------------------
# 2. DISPLAY BASIC INFORMATION
# ------------------------------------------

print("\n========== DATASET INFORMATION ==========")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 records:")
print(df.head())


# ------------------------------------------
# 3. FIND SALARY COLUMN
# ------------------------------------------

salary_column = None

possible_salary_columns = [
    "Salary",
    "salary",
    "Annual Salary",
    "annual_salary",
    "MonthlyIncome",
    "Monthly Income"
]

for column in possible_salary_columns:
    if column in df.columns:
        salary_column = column
        break


if salary_column is None:
    print("\nError: Salary column not found.")
    print("Available columns:", df.columns.tolist())
    exit()


# Convert salary column to numeric
df[salary_column] = pd.to_numeric(
    df[salary_column],
    errors="coerce"
)

# Remove rows where salary is missing
df = df.dropna(subset=[salary_column])


# ------------------------------------------
# 4. CALCULATE AVERAGE SALARY
# ------------------------------------------

average_salary = df[salary_column].mean()

print("\n========== SALARY ANALYSIS ==========")
print(f"Average Salary: {average_salary:.2f}")


# ------------------------------------------
# 5. DEPARTMENT COUNT
# ------------------------------------------

department_column = None

possible_department_columns = [
    "Department",
    "department",
    "Department Name",
    "DepartmentName"
]

for column in possible_department_columns:
    if column in df.columns:
        department_column = column
        break


if department_column is not None:

    department_count = df[department_column].value_counts()

    print("\n========== DEPARTMENT COUNT ==========")
    print(department_count)

else:
    print("\nDepartment column was not found.")


# ------------------------------------------
# 6. FILTER EMPLOYEES ABOVE SALARY THRESHOLD
# ------------------------------------------

print("\n========== SALARY FILTER ==========")

try:
    salary_threshold = float(
        input("Enter salary threshold: ")
    )

except ValueError:
    print("Invalid salary threshold.")
    exit()


high_salary_employees = df[
    df[salary_column] > salary_threshold
]

print(
    f"\nEmployees earning more than "
    f"{salary_threshold:.2f}:"
)

print(high_salary_employees)


# ------------------------------------------
# 7. EXPORT FILTERED RESULTS
# ------------------------------------------

high_salary_employees.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    f"\nFiltered employee data exported successfully "
    f"to '{OUTPUT_FILE}'."
)


# ------------------------------------------
# 8. EXPORT DEPARTMENT COUNT
# ------------------------------------------

if department_column is not None:

    department_output = "department_count.csv"

    department_count_df = (
        df[department_column]
        .value_counts()
        .reset_index()
    )

    department_count_df.columns = [
        department_column,
        "Employee Count"
    ]

    department_count_df.to_csv(
        department_output,
        index=False
    )

    print(
        f"Department count exported to "
        f"'{department_output}'."
    )


# ------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------

print("\n========== ANALYSIS SUMMARY ==========")

print(f"Total Employees: {len(df)}")
print(f"Average Salary: {average_salary:.2f}")

if department_column is not None:
    print(f"Number of Departments: {df[department_column].nunique()}")

print(
    f"Employees above threshold: "
    f"{len(high_salary_employees)}"
)

print("\nAnalysis completed successfully!")
```
